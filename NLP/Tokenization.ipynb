```python
corpus = """Hello there, this is tokenization practical.
Where we perform examples! Other sentences.
"""
```

```python
print(corpus)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
#paragraph -> senetnces
from nltk.tokenize import sent_tokenize
```

```python
document = sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
## paragraph -> word
## sentences -> word

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(sentence))
```

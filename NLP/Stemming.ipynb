## Stemming

#### Stemming algorithms (like the Porter Stemmer) apply broad cascading rules—such as dropping "-ing", "-ed", or "-es"—without checking a dictionary.
#### May result in non-words (e.g., "arguing" → "argu").
#### It is like there are words like eating eaten eat main form is eat more examples going gone goes main form is go, so main form calls as stem.

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

## PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem("congratulation")
```

```python
stemming.stem("sitting")
```

## RegexpStemmer

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem("eating")
```

```python
reg_stemmer.stem("ingeating")
```

```python
for word in words:
    print(word+"---->"+reg_stemmer.stem(word))
```

## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstem = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowballstem.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballstem.stem("fairly"),snowballstem.stem("sportingly")
```

## Wordnet Lemmatization

##### Lemmatization technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word rather than a root stem, the output of stemming. After lemmatization, we will get a valid word that means the same thing.

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
lemmatizer.lemmatize("goes",pos='v')
```

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize("fairly",pos='v'),lemmatizer.lemmatize("sportingly",pos='v')
```
